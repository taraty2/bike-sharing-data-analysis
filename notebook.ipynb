## Menentukan Pertanyaan Bisnis

**SMART Question** adalah sebuah framework untuk merumuskan pertanyaan secara terstruktur agar memeperoleh informasi yang mendalam.

* **Spesific (Spesifik)**
  * Pertanyaan harus jelas, fokus pada sebuah topik tertentu, dan tidak bermakna ganda. Hindari pertanyaan yang terlalu luas.
    * Salah: Bagaimana cara meningkatkan penjualan?
    * Benar: Faktor apa saja yang memengaruhi penurunan penjualan produk kategori elektronik di wilayah Jakarta selama kuartal terakhir?
* **Measurable (Terukur)**
  * Pertanyaan harus bisa dijawab dengan angka atau matrix yang konkret, harus tahu apa yang akan dihitung.
    * Salah: Apakah pelanggan senang dengan layanan kita?
    * Benar: Berapa skor rata-rata Customer Satisfaction untuk layanan purna jual bulan ini dibandingkan bulan lalu?
* **Action-Oriented (Berorientasi Aksi)**
  * Hasil dari pertanyaan harus bisa memberikan arahan untuk melakukan tindakan nyata. Jika pertanyaan terjawab, stakeholder harus tahu apa langkah selanjutnya.
    * Salah: Mengapa orang suka berbelanja?
    * Benar: Fitur apa pada aplikasi yang paling sering digunakan sebelum pengguna memutuskan untuk melakukan checkout?
* **Relevant (Relevan)**
  * Hasil dari pertanyaan harus sejalan dengan tujuan utama bisnis atau masalah yang sedang dihadapi.
    * Salah: Menanyakan tentang stok gudang saat masalah utamanya adalah efektivitas kampanye media sosial.
    * Benar: Apakah kampanye iklan di Instagram memberikan Return on Ad Spend (ROAS) yang lebih tinggi dibandingkan iklan di TikTok?
* **Time-bound (Terikat Waktu)**
  * Pertanyaan harus ada batasan waktu yang jelas agar analisis memiliki konteks yang tepat.
    * Salah: Berapa banyak pengguna baru kita?
    * Benar: Berapa tingkat pertumbuhan pengguna baru secara bulanan (Month-over-Month) sepanjang tahun 2025?

**Contoh pertanyaan bisnis yang memenuhi seluruh elemen SMART**

***"Faktor apa saja yang memengaruhi penurunan conversion rate pada pengguna aplikasi Android di wilayah Jabodetabek sebesar 5% selama periode Flash Sale Maret 2026?"***

Keterangan:

- **Specific**: Fokus pada "penurunan conversion rate" untuk "aplikasi Android" di "Jabodetabek". Bukan sekadar penjualan turun.
- **Measurable**: Ada angka konkret yang ingin dianalisis, yaitu penurunan sebesar "5%".
- **Action-Oriented**: Dengan mengetahui faktor penyebabnya misalnya bug pada versi Android tertentu atau kendala logistik di Jabodetabek, tim bisa langsung melakukan perbaikan teknis atau operasional.
- **Relevant**: Penurunan konversi saat Flash Sale adalah masalah kritis bagi bisnis retail/e-commerce.
- **Time-bound**: Dibatasi pada periode spesifik "Flash Sale Maret 2026".

## **Visualization & Explanatory Analysis**

#### **Pertanyaan 1: Bagaimana perkembangan total penyewaan sepeda setiap bulan serta berapa persentase perubahan jumlah penyewaan dari bulan sebelumnya selama Januari 2011–Desember 2012?**

```python
# Mengelompokkan total penyewaan berdasarkan bulan
monthly_rentals = day_df.resample('ME', on='dteday')['cnt'].sum().reset_index()

# Menghitung persentase perubahan dari bulan sebelumnya
monthly_rentals['monthly_growth'] = monthly_rentals['cnt'].pct_change() * 100

monthly_rentals.head()
```

```python
plt.figure(figsize=(14, 6))

plt.plot(
    monthly_rentals['dteday'],
    monthly_rentals['cnt'],
    marker='o',
    linewidth=2
)

plt.title('Monthly Bike Rental Trends (2011-2012)')
plt.xlabel('Month')
plt.ylabel('Total Bike Rentals')
plt.xticks(rotation=45)
plt.grid(axis='y', linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(14, 6))

plt.bar(
    monthly_rentals['dteday'],
    monthly_rentals['monthly_growth'],
    width=20
)

plt.axhline(y=0, color='black', linewidth=0.8)
plt.title('Monthly Bike Rental Growth (2011–2012)')
plt.xlabel('Month')
plt.ylabel('Monthly Growth (%)')
plt.xticks(rotation=45)
plt.grid(axis='y', linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()
```

**Interpretasi Visualisasi:**

Jumlah penyewaan sepeda cenderung meningkat sepanjang 2011–2012 dan mencapai puncaknya pada Oktober 2012. Pertumbuhan bulanan tertinggi terjadi pada April 2012 sebesar 59,6%, sedangkan penurunan terbesar terjadi pada November 2012 sebesar 23,1%. Secara keseluruhan, penyewaan mengalami fluktuasi meskipun menunjukkan tren peningkatan.

## **Analisis Lanjutan: Pengelompokan Penyewaan Berdasarkan Suhu (Temperature Binning)**

Tujuan: Mengelompokkan data penyewaan berdasarkan rentang suhu untuk mengetahui bagaimana pola rata-rata penyewaan pada setiap kategori suhu.

```python
# Mengonversi suhu ke Celsius
day_df['temp_celsius'] = day_df['temp'] * 41

day_df[['temp', 'temp_celsius']].head()
```

```python
# Mengelompokkan suhu menggunakan teknik binning
bins = [0, 10, 20, 30, 41]
labels = ['Dingin', 'Sejuk', 'Hangat', 'Panas']

day_df['temp_category'] = pd.cut(
    day_df['temp_celsius'],
    bins=bins,
    labels=labels,
    include_lowest=True
)
```

```python
# Menganalisis penyewaan berdasarkan kategori suhu
temperature_analysis = day_df.groupby(
    'temp_category',
    observed=False
).agg({
    'cnt': ['mean', 'sum'],
    'casual': 'mean',
    'registered': 'mean'
}).round(2)

temperature_analysis
```

```python
temperature_mean = day_df.groupby(
    'temp_category',
    observed=False
)['cnt'].mean()

plt.figure(figsize=(10, 6))

plt.bar(
    temperature_mean.index.astype(str),
    temperature_mean.values
)

plt.title('Average Bike Rentals by Temperature Category')
plt.xlabel('Temperature Category')
plt.ylabel('Average Bike Rentals')
plt.grid(axis='y', linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()
```

**Insight Analisis Lanjutan: Temperature Binning**

Berdasarkan pengelompokan suhu, kategori hangat memiliki rata-rata penyewaan tertinggi sebesar 5.661,69, sedangkan kategori dingin memiliki rata-rata terendah sebesar 1.677,77. Jumlah penyewaan cenderung meningkat seiring kenaikan suhu hingga kategori hangat, kemudian sedikit menurun pada kategori panas.
